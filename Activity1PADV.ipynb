```python
numeric_columns = [
    "Age",
    "Symptom Severity (1-10)",
    "Mood Score (1-10)",
    "Sleep Quality (1-10)",
    "Physical Activity (hrs/week)",
    "Treatment Duration (weeks)",
    "Stress Level (1-10)",
    "Treatment Progress (1-10)",
    "Adherence to Treatment (%)"
]

correlation_matrix = df[numeric_columns].corr()
```

```python
features = [
    "Age",
    "Gender",
    "Diagnosis",
    "Symptom Severity (1-10)",
    "Mood Score (1-10)",
    "Sleep Quality (1-10)",
    "Physical Activity (hrs/week)",
    "Medication",
    "Therapy Type",
    "Treatment Duration (weeks)",
    "Stress Level (1-10)",
    "Outcome",
    "Treatment Progress (1-10)",
    "AI-Detected Emotional State"
]

X = df[features]

y = df["Adherence Class"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
numeric_features = [
    "Age",
    "Symptom Severity (1-10)",
    "Mood Score (1-10)",
    "Sleep Quality (1-10)",
    "Physical Activity (hrs/week)",
    "Treatment Duration (weeks)",
    "Stress Level (1-10)",
    "Treatment Progress (1-10)"
]

categorical_features = [
    "Gender",
    "Diagnosis",
    "Medication",
    "Therapy Type",
    "Outcome",
    "AI-Detected Emotional State"
]

print("Numerical features:")
for col in numeric_features:
    print("-", col)

print("\nCategorical features:")
for col in categorical_features:
    print("-", col)
```

```python
print("Missing values in numerical features:")
print(X[numeric_features].isnull().sum())

print("\nMissing values in categorical features:")
print(X[categorical_features].isnull().sum())
```

```python
for col in categorical_features:
    print("\n", "="*50)
    print(col)
    print("="*50)
    print(X[col].value_counts())
```

```python
for col in categorical_features:
    print("\n", "="*50)
    print(col)
    print("="*50)
    print(X[col].value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("Processed training data shape:", X_train_processed.shape)
print("Processed testing data shape:", X_test_processed.shape)
```

```python
print("Any missing values in processed training data:",
      np.isnan(X_train_processed).sum())

print("Any missing values in processed testing data:",
      np.isnan(X_test_processed).sum())
```

```python
feature_names = preprocessor.get_feature_names_out()

print("Number of processed features:", len(feature_names))

print("\nProcessed feature names:")
for i, feature in enumerate(feature_names, 1):
    print(i, ":", feature)
```

```python
X_train_processed_df = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_processed_df = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)

display(X_train_processed_df.head())
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("Target classes:")
print(label_encoder.classes_)

print("\nEncoded training target:")
print(y_train_encoded[:20])
```

```python
print("="*60)
print("FINAL DATA CHECK")
print("="*60)

print("Original dataset:", df.shape)
print("X:", X.shape)
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
print("Processed training features:", X_train_processed.shape[1])
print("Processed testing features:", X_test_processed.shape[1])
print("Training target:", y_train_encoded.shape)
print("Testing target:", y_test_encoded.shape)
print("="*60)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
decision_tree = DecisionTreeClassifier(
    random_state=42
)
```

```python
decision_tree.fit(
    X_train_processed,
    y_train_encoded
)

print("Decision Tree model trained successfully!")
```

```python
y_pred_dt = decision_tree.predict(X_test_processed)

print("Predictions generated successfully!")
print("First 20 predictions:")
print(y_pred_dt[:20])
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
accuracy_dt = accuracy_score(y_test_encoded, y_pred_dt)

precision_dt = precision_score(y_test_encoded, y_pred_dt)

recall_dt = recall_score(y_test_encoded, y_pred_dt)

f1_dt = f1_score(y_test_encoded, y_pred_dt)

print("Decision Tree Performance")
print("=" * 40)
print("Accuracy :", round(accuracy_dt, 4))
print("Precision:", round(precision_dt, 4))
print("Recall   :", round(recall_dt, 4))
print("F1-Score :", round(f1_dt, 4))
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test_encoded,
    y_pred_dt,
    target_names=label_encoder.classes_
))
```

```python
from sklearn.metrics import confusion_matrix

cm_dt = confusion_matrix(
    y_test_encoded,
    y_pred_dt
)

print("Decision Tree Confusion Matrix:")
print(cm_dt)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_dt,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)

plt.title("Decision Tree Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()
```

```python
decision_tree_auc = DecisionTreeClassifier(
    random_state=42
)

decision_tree_auc.fit(
    X_train_processed,
    y_train_encoded
)

dt_scores = decision_tree_auc.predict_proba(
    X_test_processed
)[:, 1]
```

```python
svm_auc = SVC(
    kernel="rbf",
    probability=True,
    random_state=42
)

svm_auc.fit(
    X_train_processed,
    y_train_encoded
)

svm_scores = svm_auc.predict_proba(
    X_test_processed
)[:, 1]
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
```

```python
fpr_dt, tpr_dt, _ = roc_curve(
    y_test_encoded,
    dt_scores
)

fpr_svm, tpr_svm, _ = roc_curve(
    y_test_encoded,
    svm_scores
)

auc_dt = roc_auc_score(
    y_test_encoded,
    dt_scores
)

auc_svm = roc_auc_score(
    y_test_encoded,
    svm_scores
)

print("Decision Tree AUC:", round(auc_dt, 4))
print("SVM AUC:", round(auc_svm, 4))
```

```python
plt.figure(figsize=(8, 6))

plt.plot(
    fpr_dt,
    tpr_dt,
    label=f"Decision Tree (AUC = {auc_dt:.3f})"
)

plt.plot(
    fpr_svm,
    tpr_svm,
    label=f"SVM (AUC = {auc_svm:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve — Decision Tree vs SVM")

plt.legend()

plt.show()
```

```python
if f1_dt >= f1_svm:
    final_model = decision_tree
    final_model_name = "Decision Tree"
else:
    final_model = svm_model
    final_model_name = "SVM"

print("Selected final model:", final_model_name)
```

```python
final_results = pd.DataFrame({
    "Model": ["Decision Tree", "SVM"],
    "Accuracy": [accuracy_dt, accuracy_svm],
    "Precision": [precision_dt, precision_svm],
    "Recall": [recall_dt, recall_svm],
    "F1-Score": [f1_dt, f1_svm],
    "AUC": [auc_dt, auc_svm]
})

final_results = final_results.round(4)

print(final_results)
```
